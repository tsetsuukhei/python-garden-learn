### Дээрх кодны тайлбар

pandas.Series.str accessor нь текст боловсруулахад ашиглагддаг.
Нэмэлт мэдээлэл, зааврыг [Python | Pandas Working With Text Data](https://www.geeksforgeeks.org/pandas/python-pandas-working-with-text-data/)

```python
df['addr_lower'] = df['Address'].str.lower()  # Address баганы бүх үсгийг жижиг үсэг болгож addr_lower баганад хадгална
df['addr_upper'] = df['Address'].str.upper()  # Address баганы бүх үсгийг том үсэг болгож addr_upper баганад хадгална

df['Address'] = df['Address'].str.upper()     # Address баганы бүх үсгийг том үсэг болгож өөрт нь хадгална

df['Address'] = df['Address'].str.strip()     # Address баганы утгын эхэнд болон төгсгөлд байгаа хоосон зайг арилгана. Жишээлбэл '  123 -р байр ' => '123 -р байр'

df['Address'] = df['Address'].str.replace('-', ' ')   # Address баганы утганд байгаа бүх '-' тэмдэгтийг хоосон зайгаар солих. Жишээлбэл '123-р байр' => '123 р байр'

df['Address'].str.contains('САНСАР|ЖУКОВ')    # Address баганы утганд 'САНСАР' эсвэл 'ЖУКОВ' гэсэн үг агуулагдаж байгаа эсэхийг шалгана. Үр дүн нь True буюу False байна
                                              # Хэрвээ энэ нөхцөл үнэн бол тухайн мөрийн утгаас 'Section' багананд 'Баянзүрх' гэсэн утга оруулах жишээний кодыг дээр харуулсан
                                              # case=False гэж бичвэл том жижиг үсэг ялгахгүй шалгана


df['Address'].str.startswith('БЗД')           # Address баганы утга 'БЗД' гэсэн үгээр эхэлж байгаа эсэхийг шалгана. Үр дүн нь True буюу False байна
df['Address'].str.endswith('ХОРООЛ')          # Address баганы утга 'ХОРООЛ' гэсэн үгээр төгсөж байгаа эсэхийг шалгана. Үр дүн нь True буюу False байна

# --------------------------------------------------
# "|" (pipe) тэмдэг гэж юу вэ?
# --------------------------------------------------
# Regex дээр "|" нь OR гэсэн утгатай
#
# "ЧД|ЧИН|CHIN" гэвэл:
#   - ЧД
#   - ЧИН
#   - CHIN
# эдгээрийн аль нэг нь байвал таарна гэсэн үг
#--------------------------------------------------
# Харин "&" (ampersand) тэмдэг нь AND гэсэн утгатай
# "ЧД&1000" гэвэл:
#   - ЧД ба 1000 гэсэн үг хоёр нь байвал таарна
# --------------------------------------------------
```

## Fuzzy matching 

SANSR
SANSAAR
SANSEER
гэх мэт ялгаатай байдлаар алдаатай бичигдсэн үгсийг бүлэглэхдээ fuzzy matching ашиглах боломжтой

```python
#!pip install rapidfuzz
# доор package import хийх үед алдаа гарвал дээрх pip install командыг ажиллуулж package-ийг суулгана уу
```

```python
from rapidfuzz import process

choices = ['САНСАР', 'ЖУКОВ', 'ЯАРМАГ', 'ЗАЙСАН']

def fuzzy_match(x):
    if pd.isna(x):
        return None
    match, score, _ = process.extractOne(x, choices)
    if score > 85:
        # 85-г өөрчилж болно
        return match
    return None

df['fuzzy_area'] = df['Address'].apply(fuzzy_match)
```

## Нэмэлт хэрэгцээтэй функцүүд

```python
df = pd.DataFrame({
    'Product_ID': [101, 102, 103, 104],
    'Brand': ['ColaX', 'Yogurty', 'IcePop', 'Snacky'],
    'Flavors': [
        'Cola, Cherry, Vanilla',
        'Strawberry / Blueberry',
        'Mango; Pineapple | Orange',
        'BBQ\nCheese\nSour Cream'
    ],
    'Price': [1.5, 2.3, 0.8, 1.2]
})
```

```python
df.head()
```

```python
# Текстийг эхлээд стандартжуулна

df['Flavors'] = df['Flavors'].str.upper().str.strip()
```

```python
df.head()
```

```python
# Flavors баганы утгыг , / ; | \n тэмдэгтүүдээр салгаж шинэ Flavor_list баганад хадгална
df['Flavor_list'] = df['Flavors'].str.split(r'[,/;|\n]')

df.head()
```

```python
# Flavor_list баганы лист болгоны элементийн дагуу шинэ мөр үүсгэж өгнө
df = df.explode('Flavor_list')
```

```python
df.head()
```

```python
# Листийн элементийн эхэнд болон төгсгөлд хоосон зай байвал арилгана
df['Flavor_list'] = df['Flavor_list'].str.strip()
```

```python
# Баганы шинэ нэр
df = df.rename(columns={'Flavor_list': 'Flavor'})
```

```python
df.head()
```

```python
# Индексийг шинэчлэх бол (0-ээс эхлээд шинээр дугаарлах)

df = df.reset_index(drop=True)
```
